Q1. What is hypothesis testing in statistics?

Ans. Hypothesis testing is a fundamental statistical inference procedure used to determine whether there is enough evidence in a sample of data to infer that a certain condition is true for the entire population. It provides a disciplined, mathematical framework for making decisions, testing assumptions, and drawing conclusions from data.

Kinds of hypothesis:

* Statistical hypothesis: An assumption or statement about a population parameter (such as the mean, variance or proportion).
* Null hypothesis: A statement that there is no significant difference, relationship, or effect between populations or variables. It represents the status quo or default position that researchers aim to test against (e.g., "The new teaching method has no effect on student grades").
* Alternative hypothesis: The statement that contradicts the null hypothesis. It represents what the researcher actually believes is true, (e.g., "The new teaching method improves student grades").
* Level of significance: The probability of rejecting the null hypothesis when it actually true (a false positive). Commonly set at 0.05 (5%) or 0.01 (1%).
* Test statistic: A standardized value calculated from sample data during a test, used to decide whether to reject the null hypothesis.
* p-value: The probability of obtaining test results at least as extreme as the results actually observed, under the assumption that the null hypothesis is correct. A smaller p-value indicates stronger evidence against the null hypothesis.

The five step procedure of hypothesis testing:

1. State the hypothesis: Formulate both the null hypothesis and alternative hypothesis. Decide whether the test is one-tailed (directional) or two-tailed (non-directional).
2. Choose the significance level: Establish the threshold for statistical significance.
3. Select the appropriate test statistic: Choose the statistical test based on data type, distribution, and sample size (e.g., Z-test, student's t-test, Chi-square test, or ANOVA).
4. Compute the test statistic and p-value: Calculate the test statistic using sample data and determine the corresponding p-value or critical region boundaries.
5. Make a decision and interpret.

Errors in hypothesis testing:

* Type I error - Rejecting the null hypothesis when it is actually true (a false positive).
* Type II error - Failing to reject the null hypothesis when it is actually false (a false negative).

Common types of statistical data:

* Z-test: Used when the sample size is large (n>30) and the population variance is known.
* Student's t-test: Used for smaller sample sizes or when the population standard deviation is unknown. Variants include one-sample, independent two-sample, and paired t-tests.
* Chi-square test: Used for categorical data to test goodness-of-fit or independence between variables.
* ANOVA (Analysis of Variance): Used to compare means across three or more independent groups simultaneously.























Q2.  What is the null hypothesis, and how does it differ from the alternative
hypothesis?

Ans. The Null Hypothesis is a statement that assumes there is no significant difference, relationship, or effect between populations, variables or experimental treatments.

* The baseline assumption: It represents the status quo, current theory, or default position. When conducting a test, researchers initially assume that the null hypothesis is true until sample data provides overwhelming evidence to the contrary.
* The "No effect" rule: It typically asserts that any observed difference or pattern in the sample data is merely due to random chance or sampling error rather than a true underlying effect.
* Formulation: It is framed using terms of equality or no change (for e.g., stating that a new drug has no effect comapred to a placebo, or that the average score of two groups is equal).

Key differences:

Core meaning:-

1. Null Hypothesis - Assumes no difference, no effect or status quo.
2. Alternative Hypothesis - Assumes a real difference, effect, or relationship exists.

Burden of proof:-

1. Null Hypothesis - Default position; assumed true initially.
2. Alternative Hypothesis - Must be proven using sample evidences.

Research Goal:

1. Null Hypothesis - Researchers typically hope to reject it.
2. Alternative Hypothesis - Researchers typically hope to support or find evidence for it.

Outcome of Test:

1. Null Hypothesis - Decision is either to reject it or fail to reject it.
2. Alternative Hypothesis - Decision is either to accept support for it or fail to find support.

Practical Example:- Medical trial for a new blood pressure medication.

* Null Hypothesis - The new medication has no effect on reducing blood pressure compared to the standard treatment (mean blood pressure change is zero).
* Alternative Hypothesis - The new medication successfully reduces blood pressure more than the standard treatment (mean blood pressure change is greater than zero).
















Q3. Explain the significance level in hypothesis testing and its role in deciding
the outcome of a test?

Ans. The significance level (commonly denoted as alpha) is a critical threshold in statistical hypothesis testing. It represents the maximum risk a researcher is willing to accept of making a Type I error-which means rejecting the null hypothesis when it is actually true (a false positive). The significance level acts as a standard of proof or a gatekeeper for statistical claims.

* The Eror Threshold: When we set a significance level, we are defining the boundary of rare events. If the probability of observing our sample data (assuming the null hypothesis is true) falls below this threshold, we deem the result "statistically significant".

* The Probability Scale: It is expressed as a decimal or percentage between 0 and 1 (or 0 percent and 100 percent).

*  Control Over False Positives: It dictates how strict our test is. A lower significance level means we demand stronger evidence before changing our minds about the status quo.

Standard values and context:

* Alpha equals 0.05 (5 percent) - The most common standard. It means there is a 5 percent chance of rejecting the null hypothesis by pure random chance when it is actually true.
* Alpha equals 0.01 (1 percent) - Used in high-stakes fields like medicine or finance where a false positive is dangerous or costly. It demands much stronger evidence.
* Alpha equals 0.10 (10 percent) - Used in exploratory research where the cost of missing a potential effect (Type I error) outweighs the cost of a false alarm.

The significance level directly shapes the distribution curve of a statistical test to define the rejection region:

* Critical values: These are specific thresholds on the test statistic scale that correspond to the chosen alpha.
* Splitting alpha: For a two-tailed test, the alpha value is split equally between both tails of the distribution. For a one-tailed test, the entire alpha is placed in a single tail.
* The zone of rejection: Any calculated test statistic that falls inisde this critical boundary triggers the decision to reject the null hypothesis.

Role in deciding outcome of a test: The significance level is the ultimate benchmark used in two primary decision-making frameworks:

* The p-value approach
The p-value (probability value) represents the actual probability of obtaining test results at least as extreme as the observed results, assuming the null hypothesis is true.

Decision rule - Compare the p-value directly against alpha.

If p-value is less than or equal to alpha - The result is statistically significant. We reject the null hypothesis in favor of the alternative.

If p-value is greater than alpha - The result is not statistically significant. We fail to reject the null hypothesis due to insufficient evidence.

* The critical value approach
The decision rule - Compare the calculated test statistic (such as a t-score or z-score) to the critical value defined by alpha.

If the test statistic falls in the rejection region - We reject the null hypothesis.

If the test statistic falls outside the rejection region - We fail to reject the null hypothesis.















Q4. What are Type I and Type II errors? Give examples of each.

Ans. In a statistical hypothesis testing, decisions are made based on sample dat rather than a complete census of the population. Because samples only provide an estimate of reality, conclusions carry a degree of uncertainty. When the sample evidence leads us to make an incorrect decision, it results in a statistical error. These are categorized into type I errors and type II errors.

1.Type I error (False Positive)
  A type I error occurs when a researcher rejects a null hypothesis that is actually true. It concludes that there is a significant effect, difference, or relationship when in reality, none exists. The observed pattern in the sample data was purely random chance or sampling variation. It is also known as "False Positive" or an alpha error.

Probability - The probability of commiting a Type I error is denoted by the significance level, alpha. If your significance level is set to 0.05, you accept a 5% risk of making a Type I error.
2.Type II error (False Negative)
  It concludes that there is no significant effect or difference when, in reality, a real effect does exist in the population. Your test failed to detect it, often due to a small sample size, high data variability, or a low-powered test. It is also known as "False Negative" or beta error.

Probability - The probability of commiting a Type II error is denoted by beta.
Statistical power - The complement of a Type II error is known as the power of the test, which represents the probability of correctly detecting a true effect when it exists.

Real world example:-

Example A - The judicial system (court trial)
            In a criminal trial, the legal framework operates similarly to hypothesis testing:

* Null Hypothesis: The defendant is innocent.
* Alternative Hypothesis: The defendant is guilty.
* Type I error (Convicting an innocent person): The jury rejects the null hypothesis (finds the defendant guilty) when the defendant is actually innocent (A false positive consequence).
* Type II error (Acquitting a guilty person): The jury fails to reject the null hypothesis (finds the defendant not guilty/ lets them go free) when the defendant is actually guilty (A false negative consequence).

Example B - Medical testing (disease screening)
            Consider a diagnostic test for a serious medical condition:

* Null Hypothesis: The patient does not have the disease (healthy).
* Alternate Hypothesis: The patient has the disease.
* Type I error (False Alarm): The test tells a healthy pateint that they have the disease. This causes unnecessary panic, stress, and follow-up invasive testing.
* Type II error (Missed Diagnosis): The test tells a sick patient that they are healthy. This is often far more dangerous, as the patient walks away without receiving vital medical treatment.











Q5. What is the difference between a Z-test and a T-test? Explain when to use
each.

Ans. The fundamental difference between a Z-test and a T-test comes down to two things, whether you know the population standard deviation and how large your sample size is.

* The Z-test: You use a Z-test when the population standard deviation is known and your sample size is large (typically when n is greater than 30). It relies on the standard normal distribution (the classic bell curve). A Z-test evaluates whether two population means are different when the variances are known and the sample size is large.

Key assumptions:

1. The data is continuous and normally distributed (or the sample size is large enough via Central Limit Theorem, usually n>30).
2. The population standard deviation is known.
3. Samples are drawn randomly and independently.

The Test statistics: It calculates a Z-score, which tells you how many standard deviations your sample mean is away from the hypothesized population mean.

When to use it: In real worls, knowing the exact standard deviation of an entire population is rare. You usually only see Z-tests in introductory statistics textbooks or in massive datasets (like millions of user interactions) where the sample size is so huge that the sample standard deviation practically equals the population standard deviation.

* The T-test: You use a T-test when the population standard deviation is unknown (which is almost always the case in real-world data) and you have to estimate it using your sample data. It relies on Student's t-distribution, which has slighlty "fatter tails" to account for the extra uncertainty that comes from not knowing the true population variance. The T-test was built specifically to handle the headache of small sample sizes and unknown population variances.

Key assumptions:

1. The data is roughly normally distributed (through t-tests are fairly robust against minor deviations if sample sizes are decent).
2. The population standard deviation is unknown.

The major variants of T-tests:


1. One-sample T-test: Compares the mean of a single sample against a known or hypothesized population mean (e.g., testing if a new batch of components has a different average weight than the target 50 grams).
2. Independent two-sample T-test: Comapres the means of two completely seperate groups (e.g., comapring test scores between a control group and an experimental group).
3. Paired sample T-test: Compares means from the same group at two different times (e.g., measuring student test scores before and after a tutoring workshop).

Why the "Fatter Tails"?
Because smaller samples carry more uncertainty, a t-distribution gives more breathing room in the tails. As your sample size grows larger and larger, the t-distribution gradually morphs and starts looking identical to the standard normal z-distribution.








```python
#Q6. Write a Python program to generate a binomial distribution with n=10 and p=0.5, then plot its histogram.
#Ans.
import random
import matplotlib.pyplot as plt

n=10  #no. of trials
p=0.5 #probability pf success
num_samples = 1000

binomial_data=[]
for i in range(num_samples):
  successes = sum(1 for i in range(n) if random.random() < p)
  binomial_data.append(successes)

plt.figure(figsize=(9, 5))
plt.hist(binomial_data, bins=range(n+2), align='left', rwidth=0.85, color='skyblue', edgecolor='black')
plt.title('Binomial Distribution (n-10, p=0.5)', fontsize=14, fontweight='bold')
plt.xlabel('Number of successes', fontsize=12)
plt.ylabel('Frequency (count)', fontsize=12)
plt.xticks(range(n+1))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#Q7. Implement hypothesis testing using Z-statistics for a sample dataset in Python. Show the Python code and interpret the results.

#sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6, 50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
 #50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9, 50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

#Ans.
''' Since a strict Z-test requires a known population standard deviation, we will assume a population standard deviation of 0.5 for this demonstration.
With a sample size of n=36 (which is greater than 30), this satisfies the conditions for a large-sample Z-test.'''

import numpy as np
from scipy import stats

sample_data=[49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6, 50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
 50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9, 50.3, 50.4, 50.0, 49.7, 50.5, 49.9]
mu_0=50.5   #hypothesized population mean (null hypothesis value)
sigma=0.5   #known population standard deviation
alpha=0.05  #significance level (5%)

data=np.array(sample_data)
n= len(data)
sample_mean= np.mean(data)

#Formula: Z = (sample_mean - mu_0)/(sigma/sqrt(n))
standard_error= sigma / np.sqrt(n)
z_stat = (sample_mean - mu_0)/standard_error

p_value= 2* (1 - stats.norm.cdf(abs(z_stat)))

print(f"--- Z-test results ---")
print(f"Sample mean: {sample_mean:.4f}")
print(f"Z-statistic:{z_stat:.4f}")
print(f"P-value: {p_value:.6f}\n")

if p_value < alpha:
 print(f"Decision: Reject the null hypothesis (p-value {p_value:.4f} < alpha {alpha})")
 print("Interpretation: There is a statistically significant difference between the sample mean and the hypothesized population mean")
else:
 print(f"Decision: Fail to reject the null hypothesis (p-value {p_value:.4f} >= alpha {alpha})")
 print("Interpretation: There is insufficient evidence to conclude that the sample mean is significantly different from the hypothesized population mean")
```

```python
#Q8. Write a Python script to simulate data from a normal distribution and calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.

#Ans.
np.random.seed(42)  #set seed reproducible results
true_mean = 50.0    #true population mean
true_std = 5.0      #true population standard deviation
n = 100
sample_data = np.random.normal(loc=true_mean, scale=true_std, size=n)
sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)

standard_error = sample_std/np.sqrt(n)
confidence_level = 0.95
degrees_of_freedom = n-1
t_critical = stats.t.ppf((1+confidence_level)/2, degrees_of_freedom)

margin_or_error = t_critical * standard_error
ci_lower = (sample_mean - margin_or_error)
ci_upper = (sample_mean + margin_or_error)

print(f"sample size: {n}")
print(f"sample mean: {sample_mean:.4f}")
print(f"95% confidence interval: ({ci_lower:.4f}, {ci_upper:.4f})")

plt.figure(figsize=(10, 5))
plt.hist(sample_data, bins=15, density=True, color='skyblue', edgecolor='black', alpha=0.6, label='Sample Data Histogram')

xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = stats.norm.pdf(x, sample_mean, sample_std)
plt.plot(x,p, 'r--', linewidth=2, label='Fitted Normal Curve')

plt.axvline(sample_mean, color='green', linestyle='-', linewidth=2, label=f'Sample Mean: {sample_mean:.2f}')
plt.axvline(ci_lower, color='purple', linestyle='--', linewidth=2, label=f'95% CI Lower: {ci_lower:.2f}')
plt.axvline(ci_upper, color='purple', linestyle='--', linewidth=2, label=f'95% CI Upper: {ci_upper:.2f}')

plt.title('Simulated Normal Data and 95% Confidence Interval for the Mean', fontsize=13, fontweight='bold')
plt.xlabel('Values', fontsize=11)
plt.ylabel('Density', fontsize=11)
plt.legend(loc='upper right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#Q9.  Write a Python function to calculate the Z-scores from a dataset and visualize the standardized data using a histogram. Explain what the Z-scores represent
#in terms of standard deviations from the mean.

#Ans.
def calculate_z_scores_and_plot(data):
  data_arr = np.array(data)
  mean_val = np.mean(data_arr)
  std_val = np.std(data_arr, ddof=1)
  if std_val==0:
    raise ValueError("Standard deviation is zero, all data points are identical")
  z_scores = (data_arr - mean_val)/std_val
  plt.figure(figsize=(10, 5))
  plt.hist(z_scores, bins=10, color='skyblue', edgecolor='black', alpha=0.8)
  plt.axvline(0, color='red', linestyle='-', linewidth=2, label='Mean(Z=0)')
  plt.axvline(1, color='orange', linestyle='--', linewidth=1.5, label='-1 Std Dev')
  plt.axvline(-1, color='orange', linestyle='--', linewidth=1.5, label='+1 Std Dev')
  plt.axvline(2, color='green', linestyle='--', linewidth=1.5, label='+2 Std Dev')
  plt.axvline(-2, color='green', linestyle='--', linewidth=1.5, label='-2 Std Dev')

  plt.title('Histogram of standardized data (Z-scores)', fontsize=14, fontweight='bold')
  plt.xlabel('Z-score (number of standard deviation from the Mean)', fontsize=11)
  plt.ylabel('Frequency', fontsize=11)
  plt.legend(loc='upper right')
  plt.grid(axis='y', linestyle='--', alpha=0.7)
  plt.show()

  return z_scores

sample_dataset=[49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6, 50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
 50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9, 50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

z_results = calculate_z_scores_and_plot(sample_dataset)
print(" Calculated Z-scores:", np.round (z_results,3))
```
